# Analyze Categorical Variables -

Description: Showing summary statistics for numeric columns to understand the data distribution.

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Chart 1: Geography
sns.countplot(data=df, x='Geography', ax=axes[0])
axes[0].set_title("Customers by Geography")
axes[0].set_xlabel("Geography")
axes[0].set_ylabel("Number of Customers")

# Chart 2: Gender
sns.countplot(data=df, x='Gender', ax=axes[1])
axes[1].set_title("Customers by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

Description: As we can see that tha chart shown as highest numbers of costumers by Geography in france.

```python
# Create counts for Geography and Gender
clustered_data = pd.crosstab(df['Geography'], df['Gender'])

# Plot clustered column chart
clustered_data.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title("Customers by Geography and Gender")
plt.xlabel("Geography")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

Description: As we can see that the chart is shown as maximum number of costumers are male and are from france.

```python
fig, axis = plt.subplots(1,2, figsize = (12,4))

sns.countplot(data=df, x='Geography', hue='Exited', ax=axis[0])
axes[0].set_title("Churn by Geography")
axes[0].set_xlabel("Geography")
axes[0].set_ylabel("Number of Customers")

sns.countplot(data=df, x='Gender', hue='Exited', ax=axis[1])
axes[1].set_title("Churn by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

Description: As we can see that the chart is shown the maximum exited costumer churn from france.

```python
geography_churn = pd.crosstab(
    df['Geography'],
    df['Exited'],
    normalize='index'
) * 100
print(geography_churn)

gender_churn = pd.crosstab(
    df['Gender'],
    df['Exited'],
    normalize='index'
) * 100
print(gender_churn)
```

Description: Examine customer churn or related variables using the operations shown in this code cell.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Geography
geography_churn.plot( kind='bar', stacked=True, ax=axes[0])

axes[0].set_title("Churn Percentage by Geography")
axes[0].set_xlabel("Geography")
axes[0].set_ylabel("Percentage")
axes[0].legend(title="Exited")

# Gender
gender_churn.plot(kind='bar', stacked=True, ax=axes[1])

axes[1].set_title("Churn Percentage by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Percentage")
axes[1].legend(title="Exited")

plt.tight_layout()
plt.show()
```

Description: As we can see that the chart shows that maximum churn percentage by Gender are male and by Geography are from france.

All Boxplots help identify potential outliers.

### Numerical Features vs Churn -

For your dataset, Exited is the target variable for churn.  
  
- Exited = 1 → Customer churned  
- Exited = 0 → Customer stayed   
  
You can calculate churn rate by:  

- Geography → churn rate by country  
- Gender → churn rate by gender  
- Age → churn rate by age groups  
---
- Tenure → churn rate by years with bank  
- CreditScore → churn rate by credit-score groups  
- Balance → churn rate by balance groups 
--- 
- NumOfProducts → churn rate by number of products  
- HasCrCard → churn rate by credit-card ownership  
- IsActiveMember → churn rate by active/inactive membership  
- EstimatedSalary → churn rate by salary groups
