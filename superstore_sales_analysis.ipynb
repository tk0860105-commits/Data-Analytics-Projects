```python
# importing the necessary libraries for data analysis and visualization.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Reading the data from a csv file named "superstore_sales_data.csv".
df = pd.read_csv("superstore_sales_data.csv",encoding="windows-1252")
df.head()
```

```python
# getting information about the DataFrame.
df.info()
```

```python
# getting a statistical summary of the dataframe,like - count, mean, standard deviation, minimum, and maximum values for each numerical column.
df.describe()
```

```python
# listing all the column names in the DataFrame.
df.columns
```

```python
# checking the data types of each column in the DataFrame, which returns a Series with the data type of each column.
df.dtypes
```

```python
# checking the shape of the DataFrame.
df.shape
```

```python
# checking for missing values in the DataFrame, which returns a Series with the count of missing values for each column.
df.isnull().sum()
```

Description: As we can see that there are not any missing value in this dataset so we not need to drop anything.

```python
#if you want to drop rows with missing values from the DataFrame, modifying the DataFrame in place.
df.dropna(inplace=True)
```

```python
# checking for duplicate rows in the DataFrame, which returns a Series indicating whether each row is a duplicate of a previous row.
df.duplicated().sum()
```

Description: as we can see that there are not any duplicate value the this dataset so we not need to remove duplicates.

```python
#if we want to remove duplicate rows from the DataFrame, modifying the DataFrame in place.
df.drop_duplicates(inplace=True)
```

```python
# converting the "Order Date" and "Ship Date" columns to datetime format, handling any errors by coercing invalid parsing to NaT (Not a Time).
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
```

```python
# calculating the profit margin for each row in the DataFrame.
df["Profit_Margin"] = (df["Profit"] / df["Sales"]) * 100
```

```python
df.head(3)
```

Description: As we can see that there is a new column as profit margin before profit column. 

```python
# to calculate and print the total number of unique customers.
print(f"Total Customers   : {df['Customer ID'].nunique()}")
print(f"Total Orders      : {df['Order ID'].nunique()}")
print(f"Total Sales       : INR {df["Sales"].sum():,.2f}")
print(f"Total Profit      : INR {df["Profit"].sum():,.2f}")
```

```python
# Converting Order Date into datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

# Creating Year, Quarter and Year_Month columns
df["Year"] = df["Order Date"].dt.year
df["Year_Quarter"] = df["Order Date"].dt.to_period("Q")
df["Year_Month"] = df["Order Date"].dt.to_period("M")
```

```python
# finding sales by year
yearly_sales = df.groupby("Year", observed=True)["Sales"].sum()

print("Total Sales by Year:")
print(yearly_sales)
```

```python
# Visualizing the yearly sales trend using a bar plot.
yearly_sales.plot(kind="bar", figsize=(8, 4))
plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales ($)")
plt.tight_layout()
plt.show()
```

Description: As we see that year 2017 is with highest sales while year 2015 is with lowest sales. 

```python
# Quarterly Sales
quarterly_sales = df.groupby(
    "Year_Quarter", observed=True
)["Sales"].sum()

print("\nQuarterly Sales:")
print(quarterly_sales)
```

```python

# to visualize the quarterly sales trend using a line plot with markers.
quarterly_sales.plot(marker="o", figsize=(10, 4))
plt.title("Quarterly Sales Trend")
plt.xlabel("Year-Quarter")
plt.ylabel("Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Description: As we can see that sales is incrizing continuously acording to each quarter for every year.

```python
# Monthly Sales
monthly_sales = df.groupby(
    "Year_Month", observed=True
)["Sales"].sum()

print("\nMonthly Sales:")
print(monthly_sales)
```

```python
# to visualize the monthly sales trend using a line plot with markers.
monthly_sales.plot(marker="o", figsize=(12, 5))
plt.title("Monthly Sales Trend")
plt.xlabel("Year-Month")
plt.ylabel("Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Description: As we can see that sales is incrizing continuously acording to each month for every year.

```python
# REGION-WISE ANALYSIS
region_analysis = (
    df.groupby("Region")[["Sales", "Profit", "Quantity"]]
      .sum()
      .sort_values("Sales", ascending=False)
)

print("\nRegion-Wise Analysis:")
print(region_analysis)
```

```python
# to visualize the sales and profit by region using a bar plot.
region_analysis[["Sales", "Profit"]].plot(kind="bar")
plt.title("Sales and Profit by Region")
plt.ylabel("Amount ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Description: As we can see that west region is with highest sales and profit while south region is with lowest sales and profit.

```python
# CATEGORY-WISE SALES AND PROFIT
category_analysis = (
    df.groupby("Category")[["Sales", "Profit", "Quantity"]]
      .sum()
      .sort_values("Sales", ascending=False)
)

print("\nCategory-Wise Analysis:")
print(category_analysis)
```

```python
# to visualize the sales and profit by category using a bar plot.
category_analysis[["Sales", "Profit"]].plot(
    kind="bar"
)
plt.title("Sales and Profit by Category")
plt.ylabel("Amount ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

Description: As we can see that technology cetegory with the highest sales and profit while furniture cetegory with the lowest sales and profit.  

```python
# TOP 10 PRODUCTS BY SALES
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .nlargest(10)
)

print("\nTop 10 Products by Sales:")
print(top_products)
```

```python
# to visualize the top 10 products by sales using a horizontal bar plot.
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Sales")
plt.xlabel("Sales ($)")
plt.tight_layout()
plt.show()
```

```python
# BOTTOM 10 PRODUCTS BY SALES
bottom_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .nsmallest(10)
)

print("\nBottom 10 Products by Sales:")
print(bottom_products)
```

```python
# to visualize the bottom 10 products by sales using a horizontal bar plot.
bottom_products.sort_values().plot(kind="barh")
plt.title("Bottom 10 Products by Sales")
plt.xlabel("Sales ($)")
plt.tight_layout()
plt.show()
```

```python
# SALES VS PROFIT ANALYSIS
sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    hue="Category",
    alpha=0.6
)
plt.title("Sales vs Profit")
plt.tight_layout()
plt.show()
```

Description: As we can see that Technology stands out as the most profitable category while Office Supplies remain low-performing.

```python
# CORRELATION HEATMAP
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

Description:  as we can see that profitability is most strongly influenced by discounts (negative impact) and sales volume (positive impact) that means discounts erode margins significantly, while boosting sales without excessive discounting is the most effective path to higher profit.
